```sql
SELECT
    title,
    condition,
    price,
    seller,
    listing_date
FROM workspace.default.marketplace_output;
```

```sql
SELECT
    title,
    condition,
    CAST(
        REPLACE(
            REPLACE(price, '₹', ''),
            ',',
            ''
        ) AS DECIMAL(10,2)
    ) AS price_inr,
    seller,
    listing_date
FROM workspace.default.marketplace_output;
```

```sql
SELECT
    COUNT(*) AS total_listings,
    MIN(
        CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DECIMAL(10,2))
    ) AS min_price_inr,
    MAX(
        CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DECIMAL(10,2))
    ) AS max_price_inr,
    AVG(
        CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DECIMAL(10,2))
    ) AS avg_price_inr
FROM workspace.default.marketplace_output;
```

```sql
CREATE OR REPLACE TABLE workspace.default.marketplace_cleaned AS
SELECT
    title,
    condition,
    CAST(
        REPLACE(
            REPLACE(price, '₹', ''),
            ',',
            ''
        ) AS DECIMAL(10,2)
    ) AS price_inr,
    seller,
    listing_date,
    listing_url,
    marketplace
FROM workspace.default.marketplace_output;
```

```sql
SELECT *
FROM workspace.default.marketplace_cleaned;
```

```sql
SELECT
    condition,
    COUNT(*) AS listing_count,
    ROUND(AVG(price_inr), 2) AS avg_price_inr,
    MIN(price_inr) AS min_price_inr,
    MAX(price_inr) AS max_price_inr
FROM workspace.default.marketplace_cleaned
GROUP BY condition
ORDER BY avg_price_inr DESC;
```

```sql
SELECT
    condition,
    COUNT(*) AS listing_count,
    ROUND(AVG(price_inr), 2) AS avg_price_inr,
    MIN(price_inr) AS min_price_inr,
    MAX(price_inr) AS max_price_inr
FROM workspace.default.marketplace_output
GROUP BY condition
ORDER BY avg_price_inr DESC;
```

```sql
SELECT
    condition,
    COUNT(*) AS listing_count,
    ROUND(AVG(CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE)), 2) AS avg_price_inr,
    MIN(CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE)) AS min_price_inr,
    MAX(CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE)) AS max_price_inr
FROM workspace.default.marketplace_output
GROUP BY condition
ORDER BY avg_price_inr DESC;
```

```sql
SELECT
    title,
    condition,
    CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE) AS price_inr,
    seller,
    listing_date,
    listing_url,
    marketplace
FROM workspace.default.marketplace_output
ORDER BY price_inr ASC;
```

```sql
SELECT
    title,
    condition,
    CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE) AS price_inr,
    seller,
    listing_date
FROM workspace.default.marketplace_output
WHERE CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE) < (
    SELECT AVG(
        CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE)
    )
    FROM workspace.default.marketplace_output
)
ORDER BY price_inr ASC;
```

```sql
SELECT
    condition,
    COUNT(*) AS listing_count,
    ROUND(
        AVG(CAST(REPLACE(REPLACE(price, '₹', ''), ',', '') AS DOUBLE)),
        2
    ) AS avg_price_inr
FROM workspace.default.marketplace_output
GROUP BY condition
ORDER BY avg_price_inr ASC;
```
